```python
import numpy as np

class SimpleEmbedder:
    """
    A simple embedding approach using keyword-based vectors.
    This works without API calls and is great for learning purposes.
    """
    
    def __init__(self, dimensions: int = 100):
        self.dimensions = dimensions
        self.vocabulary = {}
        self.vocab_id = 0
    
    def _get_or_add_word(self, word: str) -> int:
        word = word.lower().strip()
        if word not in self.vocabulary:
            self.vocabulary[word] = self.vocab_id
            self.vocab_id += 1
        return self.vocabulary[word]
    
    def embed(self, text: str) -> List[float]:
        """Create a simple bag-of-words embedding"""
        words = text.lower().split()
        embedding = [0.0] * self.dimensions
        
        for word in words:
            if word.isalnum():
                word_idx = self._get_or_add_word(word)
                embedding[word_idx % self.dimensions] += 1.0
        
        # Normalize
        norm = sum(x**2 for x in embedding) ** 0.5
        if norm > 0:
            embedding = [x/norm for x in embedding]
        
        return embedding
    
    def embed_batch(self, texts: List[str]) -> List[List[float]]:
        return [self.embed(t) for t in texts]

# For Groq embeddings (when you have API key and correct model):
class GroqEmbedder:
    """Use this when you have a working Groq API key"""
    
    def __init__(self, api_key):
        self.client = Groq(api_key=api_key)
        # Try different available models
        self.model = "text-embedding-3"  # Check docs for correct model name
    
    def embed(self, text: str) -> List[float]:
        try:
            response = self.client.embeddings.create(
                model=self.model,
                input=text
            )
            return response.data[0].embedding
        except Exception as e:
            print(f"Groq embedding failed: {e}")
            print("Falling back to simple embedding")
            return SIMPLE_EMBEDDER.embed(text)
    
    def embed_batch(self, texts: List[str]) -> List[List[float]]:
        embeddings = []
        for text in texts:
            embeddings.append(self.embed(text))
        return embeddings

# Use simple embedder for now (works without Groq)
embedder = SimpleEmbedder(dimensions=100)
print("✓ Using simple embedder (fast, no API key needed)")
```

```python
class VectorStore:
    def __init__(self):
        self.vectors = []
        self.documents = []

    def add(self, doc, embedding):
        self.vectors.append(embedding)
        self.documents.append(doc)

    def cosine_similarity(self, v1, v2):
        v1, v2 = list(v1), list(v2)  # Ensure they're lists
        if len(v1) != len(v2):
            min_len = min(len(v1), len(v2))
            v1, v2 = v1[:min_len], v2[:min_len]
        
        dot = sum(a * b for a, b in zip(v1, v2))
        mag1 = sum(a * a for a in v1) ** 0.5
        mag2 = sum(b * b for b in v2) ** 0.5
        return dot / (mag1 * mag2) if mag1 and mag2 else 0

    def search(self, query_embedding, top_k=3):
        sims = [(self.cosine_similarity(query_embedding, vec), doc) 
                for vec, doc in zip(self.vectors, self.documents)]
        sims.sort(key=lambda x: x[0], reverse=True)
        return [doc for _, doc in sims[:top_k]]

# Create embeddings and build store
vs = VectorStore()
embeddings = embedder.embed_batch([c['text'] for c in chunks])
for chunk, emb in zip(chunks, embeddings):
    vs.add(chunk, emb)

print(f"✓ Vector store built with {len(vs.documents)} chunks")
```

```python
class BrightDeskChatbot:
    FALLBACK = "I couldn't find reliable information in the knowledge base. Please contact support@brightdesk.example"

    def __init__(self, vector_store, embedder):
        self.store = vector_store
        self.embedder = embedder

    def retrieve(self, query):
        query_embed = self.embedder.embed(query)
        return self.store.search(query_embed, top_k=3)

    def answer(self, query):
        docs = self.retrieve(query)
        if not docs:
            return self.FALLBACK
        return docs[0]['text']  # Return best match

chatbot = BrightDeskChatbot(vs, embedder)
print("✓ Chatbot created and ready!")
```
